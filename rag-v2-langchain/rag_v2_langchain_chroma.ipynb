```python
import langchain
import chromadb
print("All good!")
```

```python
document_text = """
Our refund policy: Customers can request a refund within 30 days of purchase.
Refunds are processed within 5-7 business days after approval.

Shipping policy: Standard shipping takes 3-5 business days.
Express shipping takes 1-2 business days and costs extra.

Support hours: Our support team is available Monday to Friday, 9am to 6pm.
"""

print(document_text)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=20
)

chunks = splitter.split_text(document_text)

for i, chunk in enumerate(chunks):
    print(f"--- Chunk {i} ---")
    print(chunk)
    print()
```

```python
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma

embedding_model = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")

vectordb = Chroma.from_texts(
    texts=chunks,
    embedding=embedding_model
)

print("Vector database created with", len(chunks), "chunks")
```

```python
query = "How long does a refund take?"

results = vectordb.similarity_search(query, k=1)

print("Your question:", query)
print("\nClosest matching chunk:")
print(results[0].page_content)
```

```python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-1.5B-Instruct",
    device_map="auto"
)

print("Model loaded")
```

```python
def rag_answer(question):
    # Step 1: retrieve the closest chunk
    results = vectordb.similarity_search(question, k=1)
    context = results[0].page_content

    # Step 2: build a grounding prompt, same idea as your original RAG project
    prompt = f"""Answer the question using ONLY the context below.
If the answer isn't in the context, say "I don't know."

Context: {context}

Question: {question}
Answer:"""

    # Step 3: generate the answer
    output = generator(prompt, max_new_tokens=60, do_sample=False)
    return output[0]["generated_text"]

response = rag_answer("How long does a refund take?")
print(response)
```

```python
response2 = rag_answer("What is your company's stock price?")
print(response2)
```
